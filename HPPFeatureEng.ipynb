Handling the Missing Values

```python
#Numerical features
numeric_df = df.select_dtypes(include='number')
print(numeric_df.isna().sum())  #We have three features with missing values : LotFrontage, MasVnrArea, GarageYrBlt
#Categorical features
categorical_df = [feature for feature in df.columns if df[feature].dtypes == 'O']
print(df[categorical_df].isna().sum()) #We have sixteen features with missing values : Alley, MasVnrType, BsmtQual, BsmtCond, BsmtExposure,
                                       # BsmtFinType1, BsmtFinType2, FireplaceQu, GarageType, GarageFinish, GarageQual, GarageCond, PoolQC,
                                       # Fence, MiscFeature
```

```python
#Examine the numerical features
print(df['LotFrontage'].describe())
print("median  ", df['LotFrontage'].median()) # -> mean and median is almost the same but we have to take care because it depends on the
                                       #neighborhood, we can also simply fill it with median values but there are around 260 empty rows!
print(df.groupby('Neighborhood')['LotFrontage'].median())
print(df['MasVnrArea'].describe())
print("median  ", df['MasVnrArea'].median())
print(df['MasVnrArea'].value_counts()) # 861 values - 0 as we only have 8 missing values we can handle it using median
print(df['GarageYrBlt'].describe())
print("median  ", df['GarageYrBlt'].median())
print(df.groupby('GarageType')['GarageYrBlt'].median()) # Every category except for None as the GarageType has a median. If i use the
#groupby for filling NaN values, it is still gonna fill NaN which will not fill out any missing values, hence we replace it by 0.
print(df['GarageType'].value_counts()) #81 None categories and 81 missing values in GarageYrBlt
```

```python
#Handling numerical values
df['LotFrontage'] = df['LotFrontage'].fillna(df.groupby('Neighborhood')['LotFrontage'].transform('median'))
df['MasVnrArea'] = df['MasVnrArea'].fillna(df['MasVnrArea'].median())
df['GarageYrBlt'] = df['GarageYrBlt'].fillna(0)
print(df['LotFrontage'].isna().sum())
print(df['MasVnrArea'].isna().sum())
print(df['GarageYrBlt'].isna().sum())
```

```python
#Handling categorical values
missing_categorical = np.where(df[categorical_df].isna().sum() > 0) # index array(tuple) of missing categorical features
for num in missing_categorical[0]: #missing_categorical has stored the indices as a single element so we selected the 0th element
#For example: missing_cat = array([1,3],[2,2]) [1,3] is the 0th element!
  feature = categorical_df[num]
  df[feature] = df[feature].fillna("None")
print(df[categorical_df].isna().sum())
```
